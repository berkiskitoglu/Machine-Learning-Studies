# inheritence

```python
class Musician():

    def __init__(self,name):
        self.name = name
        print("musician class")

    def test1(self):
        print("test1")

    def test2(self):
        print("test2")
```

```python
halit = Musician("Halit")
```

```python
halit.name
```

```python
halit.test1()
```

```python
halit.test2()
```

```python
class MusicianPlus(Musician):

    def __init__(self,name):
        Musician.__init__(self,name)
        print("musician pluss")

    def test3(self):
        print("test3")

    # overwriting
    def test1(self):
        print("test1 test1 test1")
```

```python
furkan = MusicianPlus("Furkan")
```

```python
furkan.test1()
```

```python
furkan.name = "Furkan Baş"
```

```python
furkan.name
```

```python
furkan.test3()
```

#  polymorphism

```python
class Banana():

    def __init__(self,name):
        self.name = name
        
    def info(self):
        return f"100 calories {self.name}"

class Apple():

    def __init__(self,name):
        self.name = name
        
    def info(self):
        return f"150 calories {self.name}"
```

```python
banana = Banana("banana")
```

```python
apple = Apple("apple")
```

```python
banana.info()
```

```python
apple.info()
```

```python
fruit_list = [banana,apple]
```

```python
for fruit in fruit_list:
    print(fruit.info())
```

# encapsulation

```python
class Phone():
    
    def __init__(self,name,price):
        self.name = name
        self.__price = price

    def info(self):
        print(f"{self.name} price is {self.__price}")

    def change_price(self,price):
        self.__price = price
```

```python
iphone = Phone("iPhone 14",500)
```

```python
iphone.info()
```

```python
iphone.price = 400
```

```python
iphone.info()
```

```python
iphone.change_price(300)
```

```python
iphone.info()
```

# abstraction

```python
from abc import ABC, abstractmethod
```

```python
class Car(ABC):

    @abstractmethod
    def max_speed(self):
        pass
```

```python
#my_car = Car()
```

```python
class Tesla(Car):

    def max_speed(self):
        print("200km")
    pass
```

```python
tesla = Tesla()
```

```python
tesla.max_speed()
```

```python
class Mercedes(Car):

    def max_speed(self):
        print("200km")
```

```python
mercedes = Mercedes()
```

```python
mercedes.max_speed()
```

```python
class Fruit():

    def __init__(self,name,calories):
        self.name = name
        self.calories = calories
        
    def __str__(self):
        return f"{self.name}: {self.calories} calories"

    def __len__(self):
        return self.calories
```

```python
myFruit = Fruit("banana",150)
```

```python
myFruit.calories
```

```python
myFruit.name
```

```python
print(myFruit)
```

```python
len(myFruit)
```

```python
class Train():

    def __init__(self,name):
        self.name = name
    def __getitem__(self,key):
        if key == "a":
            return self.name
        else:
            return "Not Found"
```

```python
myTrain = Train("myTrain")
```

```python
myTrain["a"]
```

```python
# args , kwargs (arguments, key word arguments)
```

```python
def arg_sum(*args):
    return sum(args)
```

```python
arg_sum(3,2,1,51,21,2)
```

```python
def args_example(*args):
    print(args)
```

```python
args_example(1231,141,151,11,452)
```

```python
def kwargs_example(**kwargs):
    print(kwargs)
```

```python
kwargs_example(apple=100,banana=100,melon=200)
```

```python
def kwargs_example_2(**kwargs):
    if "apple" in kwargs:
        print("applee")
    else:
        print(":(")
```

```python
kwargs_example_2(apple=10,banana=20)
```

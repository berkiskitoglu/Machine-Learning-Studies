```python
app = ['GAME','COMMUNICATION','TOOLS','PRODUCTIVITY','SOCIAL']
plt.figure(figsize=(40,30))
for i,app in enumerate(app):
    df2 = df_app_category[df_app_category.Category == app]
    df2 = df2.head(5)
    #print(df2)
    plt.subplot(3,2,i+1)
    sns.barplot(data=df2,x='Installs',y='App')
    plt.title(app,size=20)
plt.tight_layout()
plt.show()
```

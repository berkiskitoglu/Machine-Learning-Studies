```python
def salary_category(salary):
    if salary < 50000:
        return 'Low'
    elif 50000 <= salary < 80000:
        return 'Medium'
    else:
        return 'High'
```

```python
df['Salary'].apply(salary_category)
```

```python
df['Salary_Category'] = df['Salary'].apply(salary_category)
```

```python
# if experience > 10  => performance_score + 1
```

```python
def add_score(experience,performance_score):
    if experience > 10:
        performance_score += 1
        return performance_score
    else:
        return performance_score
```

```python
def adjust_performance(experience):
    if experience > 10:
        return 1
    else:
        return 0
```

```python
df['Adjusted'] = df['Experience'].apply(adjust_performance)
df
```

```python
df['NewScore'] = df['Performance_Score'] + df['Adjusted']
```

```python
df
```

```python
def adjust_new(row):
    if row['Experience'] > 10:
        return row['Performance_Score'] + 1
    else:
        return row['Performance_Score']
```

```python
df['Adjusted_Score'] = df.apply(adjust_new,axis=1)
```

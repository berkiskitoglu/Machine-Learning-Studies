```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Görsellrr oluştururken jupyter notebook üzerinde kolay bir şekilde göstermemize olanak tanıyor . Sadece notebook için
%matplotlib inline
```

```python
age_list = [10,20,30,30,30,40,50,60,70,75]
weight_list = [20,60,80,85,86,87,70,90,95,90]
```

```python
plt.plot(age_list,weight_list,'y')
plt.title("Age vs Weight")
plt.xlabel('Age')
plt.ylabel('Weight')
plt.show()
```

```python
# numpy
```

```python
np_age_list = np.array(age_list)
np_weight_list = np.array(weight_list)
```

```python
plt.plot(np_age_list,np_weight_list,'g')
plt.xlabel('Age')
plt.ylabel('Weight')
plt.title('Age vs Weight')
plt.show()
```

```python
# linespace => başlangıç değerinden verilen değer aralığında x tane değer al ve bunların arasındaki artış miktarı eşit olsun
numpy_arr1 = np.linspace(0,10,20)
numpy_arr1
```

```python
# subplot => row , column, x.grafik / yan yana grafik çizdirmek için kullanılır
plt.subplot(1,2,1)
plt.plot(numpy_arr1,numpy_arr2,'r*-')
plt.subplot(1,2,2)
plt.plot(numpy_arr2,numpy_arr1,'r--')
plt.show()
```

```python
# figure => özelleştirmek için kendi figure ' müzü oluşturuyoruz 
# figure.add_axes([0,1,0.1,0.3,0.3]) => buradaki ilk iki değer diğer grafikler eklendikten sonraki boşluklar diğer ikisi genişilik ve yükseklik 
my_figure = plt.figure()
figure_axes = my_figure.add_axes([0.1,0.1,0.7,0.7])
figure_axes.plot(numpy_arr1,numpy_arr2,'g')
figure_axes.set_xlabel('X Axis')
figure_axes.set_ylabel('Y Axis')
figure_axes.set_title('Graph Title')
plt.show()
```

```python
# dpi => çözünürlük
new_fig = plt.figure(dpi=100)
new_axes = new_fig.add_axes([0.1,0.1,0.9,0.9])
new_axes.plot(numpy_arr1 ,numpy_arr2 ** 2 , label = 'numpy array ** 2 ')
new_axes.plot(numpy_arr1 ,numpy_arr2 ** 3 , label = 'numpy array ** 3 ')
new_axes.legend(loc=6)
plt.show()
```

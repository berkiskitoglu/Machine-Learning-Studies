```python
# merge
```

```python
df_merge_1 = pd.read_csv("C:/Users/iskit/Desktop/Machine-Learning-Studies/02-DataScience/datasets/merge_data1.csv")
df_merge_2 = pd.read_csv("C:/Users/iskit/Desktop/Machine-Learning-Studies/02-DataScience/datasets/merge_data2.csv")
```

```python
df_merge_1
```

```python
df_merge_2
```

```python
# merge - outer join
```

```python
df_merged_outer = pd.merge(df_merge_1,df_merge_2,on='Employee_ID',how="outer")
df_merged_outer
```

```python
# merge - left join
```

```python
df_merged_left = pd.merge(df_merge_1,df_merge_2,on="Employee_ID",how='left')
df_merged_left
```

```python
# merge - right join
```

```python
df_merged_right = pd.merge(df_merge_1,df_merge_2,on="Employee_ID",how='right')
df_merged_right
```

```python
# merge - inner join
```

```python
df_merged = pd.merge(df_merge_1,df_merge_2,on='Employee_ID',how='inner')
df_merged
```
